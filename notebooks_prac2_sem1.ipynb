```python
                                                ## ASSIGNMENT NO. 2

# Write a python program to find the largest of
# three numbers while implementing different forms of
# if..else statement.
a = int(input("Enter first number: "))
b = int(input("Enter second number: "))
c = int(input("Enter third number: "))

if a >= b and a >= c:
    largest = a
elif b >= a and b >= c:
    largest = b
else:
    largest = c

print("The largest number of the given three numbers is:", largest)
```
